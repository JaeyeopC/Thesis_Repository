```python
FILE_PATH = 'data/stepverify_labeled_0.9.json'
TRAIN_OUT_PATH = 'train_test_split/train_stepverify_labeled_0.9.json'
TEST_OUT_PATH = 'train_test_split/test_stepverify_labeled_0.9.json'
```

```python
import json
from sklearn.model_selection import train_test_split
import pandas as pd

def load_data(file_path=FILE_PATH):
    with open(file_path, 'r') as f:
        all_data = json.load(f)

    return all_data

def split_train_test(data, stratify_col="error_category", test_size=0.3, random_state=42):
    # stratify에 사용할 label
    stratify_labels = [item[stratify_col] for item in data]
    indices = list(range(len(data)))

    train_data, test_data, train_idx, test_idx = train_test_split(
        data,
        indices,
        test_size=test_size,
        random_state=random_state,
        stratify=stratify_labels,
    )

    return train_data, test_data, train_idx, test_idx
```

```python
all_data = load_data(FILE_PATH)

train, test, train_idx, test_idx = split_train_test(all_data)


print(f"Before split length: {len(all_data)} | label counts: \n")
all_counts = pd.Series([item["error_category"] for item in all_data]).value_counts()
print(all_counts)

print("-" * 100)

print(f"Total Train Length: {len(train)} | Train label counts: \n")
train_counts = pd.Series([item["error_category"] for item in train]).value_counts()
print(train_counts)

print("-" * 100)

print(f"Total Test Length: {len(test)} | Test label counts: \n")
test_counts = pd.Series([item["error_category"] for item in test]).value_counts()
print(test_counts)
```

```python
os.makedirs(os.path.dirname(TRAIN_OUT_PATH), exist_ok=True)

# save train
with open(TRAIN_OUT_PATH, 'w', encoding='utf-8') as f:
    json.dump(train, f, ensure_ascii=False, indent=4)

# save test
with open(TEST_OUT_PATH, 'w', encoding='utf-8') as f:
    json.dump(test, f, ensure_ascii=False, indent=4)


print(f"Train saved: {TRAIN_OUT_PATH} ({len(train)} samples)")
print(f"Test saved:  {TEST_OUT_PATH} ({len(test)} samples)")
```
